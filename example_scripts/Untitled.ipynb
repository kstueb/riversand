```python
# chi2.pdf same as matlab pdf
fig, ax = plt.subplots(figsize=(4,3))
x = np.linspace(0,15,100)
for df in [1,2,3,4]:
    y = stats.chi2.pdf(x,df)
    ax.plot(x,y, label="df={}".format(df))
ax.set_xlabel("x")
ax.set_ylabel("pdf")
plt.legend()
plt.grid()
```

```python
# chi2.cdf same as matlab cdf
fig, ax = plt.subplots(figsize=(4,3))
x = np.linspace(0,15,100)
for df in [1,2,3,4]:
    y = stats.chi2.cdf(x,df)
    ax.plot(x,y, label="df={}".format(df))
ax.set_xlabel("x")
ax.set_ylabel("cdf")

plt.legend()
plt.grid()
```

```python
# chi2.cdf same as matlab cdf
fig, ax = plt.subplots(figsize=(4,3))
x = np.linspace(0,15,100)
for df in [1,2,3,4]:
    y = stats.chi2.ppf(x,df)
    ax.plot(x,y, label="df={}".format(df))
ax.set_xlabel("x")
ax.set_ylabel("cdf")

plt.legend()
plt.grid()
```

```python
# Taylor:
Tchi2 = np.array([0  ,.5,1 ,1.5, 2,2.5,  3,3.5,  4,4.5,  5,5.5,6,8,10])

Td1 =   np.array([100,48,32, 22,16, 11,8.3,6.1,4.6,3.4,2.5,1.9,1.4,.5,.2])/100
Td2 =   np.array([100,61,37, 22,14,8.2,  5,  3,1.8,1.1,0.7,0.4,0.2])/100
Td3 =   np.array([100,68,39, 21,11,5.8,2.9,1.5, .7, .4, .2, .1])/100

tx1 = Tchi2[:(len(Td1))]
tx2 = Tchi2[:(len(Td2))]
tx3 = Tchi2[:(len(Td3))]

fig, ax = plt.subplots(figsize=(4,4))
ax.plot(tx1*1,1-Td1, '.-', label="df=1")
ax.plot(tx2*2,1-Td2, '.-', label="df=2")
ax.plot(tx3*3,1-Td3, '.-', label="df=3")

for df in [1,2,3]:
    ax.plot(Tchi2, stats.chi2.cdf(Tchi2, df))
ax.grid()
```

```python

xx = np.linspace(0,30, 500) # reduced chi2; xx = chi2/df

fig, ax = plt.subplots(figsize=(5,3))
for df in [1,2,3,4]:
    yy = stats.chi2.cdf(xx, df) # cdf of chi2
    ax.plot(xx/df,yy,label="df="+str(df)) # x-axis rescaled to represent reduced chi2 = chi2/df

ax.set_xlabel("chi2")
ax.set_ylabel("p")
ax.set_xlim((0, 1.5))
ax.legend()
ax.grid()

p = 0.5
df = 3

chi2 = stats.chi2.ppf(p,df)
_ = ax.plot(chi2/df, p, '*k')
```

```python
xx = np.linspace(0,10, 100) # reduced chi2; xx = chi2/df

fig, ax = plt.subplots(figsize=(4,4))
for df in [1,2,3,4]:
    yy = stats.chi2.cdf(xx*df, df)
    ax.plot(xx,yy,label=str(df))

ax.set_xlabel("chi2")
ax.set_ylabel("p")
ax.grid()
```
