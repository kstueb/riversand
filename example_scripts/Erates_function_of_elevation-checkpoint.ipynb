## Erosion rates

Based on `BalcoCronus3_ErosionRate.jpyn` to demonstrate how production rate depends on elevation.

This is a script to run the `BalcoCronus3` erosion rate calculator in the most standard way.

It returns erosion rates and uncertainties for St, Lm, and LSDn scaling for a given site for either Be-10 or Al-26. There are also some examples how to get Rc-time series or generate plots of Rc, scaling factors, or production rates.

````python
# see Erosion_rates_from_Excel.ipynb for reading data from a spread sheet

lat       = 37.49912     # site latitude, dec. degr.
lon       = 71.60012     # site longitude, dec. degr. 
elev      = 2086.     # site elevation
press     = 'ERA40'       # site atm. pressure, hPa or 'ERA40'
thick     = 0        # sample thickness, cm
rho       = 2.65     # density, g/cm3
shielding = 1        # shielding
#erate     = 0        # erosion rate, cm/yr
yr        = 2017     # calendar year of sampling
N10       = 10971100   # nuclide concentration, at/g
delN10    = 4702
N26       = 705325   # nuclide concentration, at/g
delN26    = 30097
````

````python
# translated from matlab original v3.0.2
# https://hess.ess.washington.edu/math/v3/

import numpy as np
import matplotlib.pyplot as plt
````

````python
import sys
#sys.path.append('/home/user/path/to/folder')
sys.path.append('/home/user/jupyter/calculators') # Lenovo
#sys.path.append('/home/stuebner/python_playground/jupyter/CRN_Calculators') # Dell

import BalcoCronus3
````

````python
## use getXX_sf() to explore production rate time series;
# this is independent of the input nuclide concentration
fig, ax = plt.subplots(figsize=(8,6))
for elev in [2120, 3000, 4000, 5000, 6600]:
    s = BalcoCronus3.sample(lat=lat, lon=lon, elev=elev, press='ERA40', 
                        shielding=shielding, thick=thick, rho=rho,
                        N10=N10, delN10=delN10,
                        N26=N26, delN26=delN26,
                        yr=yr)
    
    s.set_pressure()
    
    _ = s.get_exposure_age_St(nuclide='Be')
    n = s.Be
    #Rc = BalcoCronus3.get_LiftonRc(s, t_max=2005000, plotting=False)
    Rc = BalcoCronus3.get_LiftonRc(s, t_max=200000, plotting=False)
    Rc.nuclide = n.nuclide
    Rc.pressure = s.press
    Rc.sf_Lm = BalcoCronus3.getLm_sf(Rc, plotting=False)
    
    # Production rates:
    Psp_Lm = s.shielding * n.thick_sf * n.P0_Lm * Rc.sf_Lm
    if elev in [3000, 4000, 5000]:
        plt.plot(Rc.tmax/1e3, Psp_Lm, linewidth='1', color='#1f77b4', label=str(elev)+' m')
    else:
        plt.plot(Rc.tmax/1e3, Psp_Lm, linewidth='1', color='#ff7f0e', label=str(elev)+' m')
#plt.title('Spallation production rates, Lm')
plt.legend()
plt.xlabel('time, kyr', fontsize=16, labelpad=10)
plt.ylabel('Psp (Lm), at/g/yr', fontsize=16, labelpad=10)
plt.tick_params(axis='both', which='major', labelsize=16)
plt.xticks(ticks=[0,50,100,150,200])
plt.show()
````
