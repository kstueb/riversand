# From a DEM to catchmentwide erosion rates

## DEM preprocessing and catchment delineation:
using QGIS with the SAGA toolbox

1. load the DEM into a QGIS project, reproject into utm if necessary, note the EPSG
2. from the processing Toolbox, apply SAGA <font color=1D57A9 >'Fill sinks (wang & liu)'</font>
3. The GPS coordinates of the sample site may not lie exactly on the channel network. Extract the channel network and correct any offset of your sampling site:
4. use SAGA <font color=1D57A9 >'Strahler order'</font> to create a strahler order raster<br> or use SAGA <font color=1D57A9 >'Channel network and drainage basins'</font> to create a channels-vector file
6. note the UTM coordinates of your (corrected) sampling site
7. use SAGA <font color=1D57A9 >'Upslope area'</font> with target coordinates from pour points and filled elevation data <br>
and polygonize the resulting raster (<font color=1D57A9 >Raster>Conversion>Polygonize</font>); save as shapefile
8. delete from the shapefile all polygons except the one that represents your catchment and save
9. perform a validity check (<font color=1D57A9 >'Vector>Geometry tools>Check validity'</font>) and fix any errors that may have been found

### How to deal with missing projection in raster data
A geotiff is a tif with additional information about its spatial location. This information includes the Coordinate Reference System (crs) such as Lat/Long or UTM zone. Information about the crs may be missing for some raster datasets, for example if the topographic shielding was calculated with Matlab without the Mapping toolbox. This program will try to deal with this by assuming that the crs is identical to the DEM, but it is safer to provide geotiffs with crs specified. The crs can be set with software such as `gdal`, `ArcGIS` or `QGIS`. You can also use the code in the following cell (function `write_epsg_to_geotiff()`) to create a copy of your raster dataset with the appropriate crs; this requires that the raster is projected and that you know the epsg code of the projection (see https://spatialreference.org/)!!

## 1. Specify input data

````python
# Specify input data:
path_to_data = './user_data' # local

# Raster datasets are defined by definitions of label & filename:
# The DEM must be labelled 'Elevation'.
# If a topographic shielding raster is specified it should be labelled 'Shielding'.
dfg = None
dfg = pd.DataFrame(columns={'fname', 'label', 'src'}) # initialize dataframe of raster datasets

dfg.loc[len(dfg)] = {'label': 'Shielding',  'fname': 'test_shielding.tif'}
dfg.loc[len(dfg)] = {'label': 'Elevation',  'fname': 'test_dem.tif'}
dfg.loc[len(dfg)] = {'label': 'Slope',      'fname': 'test_slope.tif'}

# Optional: use constant shielding factor instead of a shielding raster dataset
#constant_toposhielding = 0.99

# Catchment outline shapefile:
fname_catchment = 'test_catchment.shp'

# Optional: resample all raster datasets to predefined resolution, either scalar x=y or tuple (x, y)
#resample_res = (100.0, 100.0) 

# Other options:
resample_raster_if_necessary = True # set to False if all rasters have identical resolution and aligned pixels
plotting = True # set to False to skip graphic output

if 'Elevation' not in set(dfg.label): print("\nERROR: An Elevation raster must be provided!")
````

## 2. Import / validate the spatial data

````python
validate_path(path_to_data, param_summary)

# Get handle 'src' to each raster
for index, raster in dfg.iterrows():
    dfg.loc[index,'src'] = validate_raster(path_to_data, raster['fname'], raster['label'], param_summary)
    
# Validate that all rasters have the same projection
epsg = None
for index, raster in dfg.iterrows():
    dfg.loc[index,'src'], epsg = validate_raster_crs(raster['src'], raster['label'], epsg)
    
# Validate the catchment shapefile
polygon = validate_catchment(path_to_data, fname_catchment, param_summary, epsg)
````

````python
# generate some summary output
crs = CRS.from_epsg(epsg)
print("{}  (epsg:{})".format(crs.name, epsg))
print("\nArea of use: {}".format(crs.area_of_use.name))
print("\nUnits: {}, {}".format(crs.axis_info[0].unit_name, crs.axis_info[1].unit_name))
````

````python
# resolution of the Elevation raster
DEM_res = add_resolution(dfg.loc[dfg[dfg['label']=='Elevation'].index[0],'src'].res, CRS.from_epsg(epsg), param_summary)

# resolution to resample all rasters to (either resolution of Elevation raster or specified in the input)
if resample_raster_if_necessary and (resample_res is None):
    resample_res = dfg.loc[dfg[dfg['label']=='Elevation'].index[0],'src'].res # get the resolution of the elevation raster

# resample raster to resolution of DEM (if necessary) or to specified resolution 'resample_res'
if resample_res is not None:
    for index, raster in dfg.iterrows():
        dfg.loc[index,'src'] = resample_raster(raster['src'],
                                               dst_res=resample_res,
                                               param_summary=param_summary)

# verify that all raster datasets have the same resolution
verify_resolutions([r['src'] for _,r in dfg.iterrows()])

# sort the raster dataset such that 'Elevation' is the first one; necessary for elevation binning
dfg.reset_index(inplace=True, drop=True)
dfg['temp'] = range(len(dfg), 2*len(dfg))
dfg.loc[dfg.index==dfg[dfg['label']=='Elevation'].index[0], 'temp'] = 0
dfg = dfg.sort_values('temp').drop('temp', axis=1)
dfg.reset_index(inplace=True, drop=True)
````

````python
# clip the catchment in all raster datasets
XX = [clip_raster(raster['src'], polygon, raster['label']) for index, raster in dfg.iterrows()]

print("Raster resolution is {}".format(dfg.loc[0,'src'].res))
````
